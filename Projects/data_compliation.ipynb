# Setup

```python
import pandas as pd
import torch
import torch.nn as nn
import util
import os
```

## Read in the Data

```python
# Get all directories in the specified path

# Change this to point to the gene_expression_quantification folder. This script will pull all the file paths from this directory.
parent_dir = 'data/gene_expression_quantification'
directory_names = [name for name in os.listdir(parent_dir) 
                  if os.path.isdir(os.path.join(parent_dir, name))]

print(f"Found {len(directory_names)} directories")
print(directory_names[:5]) # Print first 5 directories as a sample
path_names = []

for directory_name in directory_names:
    file_name = os.listdir('data/gene_expression_quantification/' + directory_name)
    path_name = 'data/gene_expression_quantification/' + directory_name + '/' + file_name[0]
    path_names.append(path_name)

print(path_names[:5])
```

# Processing

## Compile Data into a single Dataframe

```python
df = pd.read_csv(path_names[0], sep='\t', header=1)[['gene_name','unstranded']]
df
```

```python
compiled_df = pd.DataFrame()
counter = 0
for path in path_names:
    df = pd.read_csv(path_names[0], sep='\t', header=1)[['gene_name','unstranded']] # These are the columns to keep
    counter += 1
    print(f"Processing file {counter}")
    compiled_df = pd.concat([compiled_df, df], axis=0)
print("Number of rows in compiled DataFrame:", len(compiled_df))
compiled_df.head(5)    
```

## Export Data

```python
compiled_df.to_csv('data/compiled_gene_expression.csv', index=False)
compiled_df.to_pickle('data/compiled_gene_expression.pkl')
```
